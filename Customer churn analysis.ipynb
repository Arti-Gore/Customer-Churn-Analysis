```python
import matplotlib.pyplot as plt

pd.crosstab(df['gender'], df['Churn']).plot(kind='bar')
plt.title('Gender-wise Customer Churn')
plt.xlabel('Gender')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.show()
```

```python
pd.crosstab(df['Contract'], df['Churn']).plot(kind='bar')
plt.title('Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.show()
```

```python
sns.boxplot(x='Churn', y='MonthlyCharges', data=df)
plt.title('Monthly Charges vs Customer Churn')
plt.xlabel('Churn')
plt.ylabel('Monthly Charges')
plt.show()
```

```python
sns.boxplot(x='Churn', y='tenure', data=df)
plt.title('Customer Tenure vs Churn')
plt.xlabel('Churn')
plt.ylabel('Tenure (Months)')
plt.show()
```

```python
sns.boxplot(x='Churn', y='tenure', data=df)
plt.title('Customer Tenure vs Churn')
plt.xlabel('Churn')
plt.ylabel('Tenure (Months)')
plt.show()
```

```python
pd.crosstab(df['PaymentMethod'], df['Churn']).plot(kind='bar', figsize=(10,5))
plt.title('Churn by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.xticks(rotation=30)
plt.show()
```

```python
pd.crosstab(df['SeniorCitizen'], df['Churn']).plot(kind='bar')
plt.title('Churn by Senior Citizen Status')
plt.xlabel('Senior Citizen (0 = No, 1 = Yes)')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.show()
```

```python
df['TenureGroup'] = pd.cut(
    df['tenure'],
    bins=[0, 12, 24, 48, 72],
    labels=['0-12 Months', '13-24 Months', '25-48 Months', '49-72 Months']
)

pd.crosstab(df['TenureGroup'], df['Churn']).plot(kind='bar', figsize=(9,5))
plt.title('Churn by Customer Tenure Group')
plt.xlabel('Tenure Group')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.xticks(rotation=0)
plt.show()
```

```python
pd.crosstab(df['PaperlessBilling'], df['Churn']).plot(kind='bar')
plt.title('Churn by Paperless Billing')
plt.xlabel('Paperless Billing')
plt.ylabel('Number of Customers')
plt.legend(title='Churn')
plt.show()
```

```python
contract_churn_rate = pd.crosstab(
    df['Contract'],
    df['Churn'],
    normalize='index'
) * 100

print(contract_churn_rate)
```

```python
# Payment Method Churn Rate
payment_churn_rate = pd.crosstab(
    df['PaymentMethod'],
    df['Churn'],
    normalize='index'
) * 100

print(payment_churn_rate.round(2))
```

```python
internet_churn_rate = pd.crosstab(
    df['InternetService'],
    df['Churn'],
    normalize='index'
) * 100

print(internet_churn_rate.round(2))
```

# Key Insights

1. Customers with month-to-month contracts have a higher churn rate.

2. Customers with higher monthly charges are more likely to churn.

3. New customers with lower tenure show a higher probability of churn.

4. Fiber optic internet service customers show relatively higher churn.

5. Electronic check payment method has a higher churn rate.

6. Senior citizens show a comparatively higher churn rate.

7. Long-term contracts and longer customer tenure are associated with lower churn.

   

# Business Recommendations

1. Encourage customers to choose long-term contracts by offering discounts and special benefits.

2. Provide attractive offers and onboarding support to new customers during their initial months.

3. Review pricing strategies for customers with high monthly charges.

4. Improve service quality and technical support for high-risk customer segments.

5. Promote automatic payment methods to improve customer convenience.

6. Create special customer retention programs for senior citizens.

7. Identify high-risk customers early and provide personalized offers to reduce churn.

# Conclusion

This Customer Churn Analysis project analysed customer data to identify the major factors influencing customer churn.The analysis revealed that contract type, customer tenure, monthly charges, internet service, payment method, and senior citizen status are important factors associated with customer churn.
Customers with month-to-month contracts, shorter tenure, and higher monthly charges showed a higher tendency to churn.The insights obtained from this analysis can help businesses identify high-risk customers and implement effective retention strategies to improve customer satisfaction and reduce churn.
